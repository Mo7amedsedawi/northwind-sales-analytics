```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
BASE_DIR = Path.cwd().parent
RAW_DATA_DIR = BASE_DIR / "data" / "raw"

print(RAW_DATA_DIR)
```

```python
orders = pd.read_csv(RAW_DATA_DIR / "Orders.csv")
order_details = pd.read_csv(RAW_DATA_DIR / "Order Details.csv")
products = pd.read_csv(RAW_DATA_DIR / "Products.csv")
customers = pd.read_csv(RAW_DATA_DIR / "Customers.csv")
employees = pd.read_csv(RAW_DATA_DIR / "Employees.csv")
```

```python
order_details["Quantity"] = pd.to_numeric(
    order_details["Quantity"],
    downcast="integer"
)

order_details["ProductID"] = pd.to_numeric(
    order_details["ProductID"],
    downcast="integer"
)

order_details["OrderID"] = pd.to_numeric(
    order_details["OrderID"],
    downcast="integer"
)
```

```python
orders["OrderDate"] = pd.to_datetime(
    orders["OrderDate"],
    errors="coerce"
)

orders["RequiredDate"] = pd.to_datetime(
    orders["RequiredDate"],
    errors="coerce"
)

orders["ShippedDate"] = pd.to_datetime(
    orders["ShippedDate"],
    errors="coerce"
)
```

```python
orders.info()
```

```python
sales = orders.merge(
    order_details,
    on="OrderID",
    how="inner"
)

sales.head()
```

```python
sales = sales.merge(
    products,
    on="ProductID",
    how="left"
)

sales.head()
```

```python
sales = sales.merge(
    customers,
    on="CustomerID",
    how="left"
)

sales.head()
```

```python
sales = sales.merge(
    employees,
    on="EmployeeID",
    how="left"
)

sales.head()
```

```python
sales.info()
```

```python
sales['Revenue'] = sales['UnitPrice_x'] * sales['Quantity'] * (1 - sales['Discount'])
```

```python
customer_sales = (
    sales
    .groupby(["CustomerID", "CompanyName"], as_index=False)
    ["Revenue"]
    .sum()
    .sort_values("Revenue", ascending=False)
)

customer_sales.head(10)
```

```python
employee_sales = (
    sales
    .groupby(
        ["EmployeeID", "FirstName_x", "LastName_x"],
        as_index=False
    )["Revenue"]
    .sum()
    .sort_values("Revenue", ascending=False)
)

employee_sales.head(10)
```

```python
sales["Year"] = sales["OrderDate"].dt.year

yearly_sales = (
    sales
    .groupby("Year", as_index=False)["Revenue"]
    .sum()
)

yearly_sales
```
